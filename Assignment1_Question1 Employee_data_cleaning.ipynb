```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank'],
    'Age': [25, np.nan, 30, 22, np.nan, 35],
    'Salary': [50000, 60000, np.nan, 45000, np.nan, 70000],  # 6 items (replace/adjust values as needed)
    'Department': ['HR', 'IT', 'IT', np.nan, 'HR', 'Finance']
}

df = pd.DataFrame(data)
print("Original data")
print(df)
```

```python
df_renamed = df.rename(columns= {
    "Name" : "Ëmployee_Name",
    "Salary": "Annual_Sal",
    "Department": "Dept"
})
print(df_renamed)
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
df[df['Department'] == 'IT']
print(df)
```

```python
df[df['Salary'] > 60000]
print(df)
```

```python
df['Salary'].mean()
print(df)
```

```python
df.groupby('Department')['Salary'].mean()
print(df)
```

```python
# Task 1 Extraction

import pandas as pd
```

```python
# Import the csv file
raw_df = pd.read_csv("employee_salary.csv")
print(f"Raw Data: {raw_df.shape[0]} rows, {raw_df.shape[1]} columns")
```

```python
# Task 2 - Transformation
df = raw_df.drop_duplicates()
```

```python
# Select relevant columns
df = df[["Name","Age","Department","Monthly_Salary"]]
```

```python
df = df[(df["Monthly_Salary"].notna()) & (df["Age"] >=18)]
```

```python
df = df.rename(columns={
    "Name": "Employee",
    "Department": "Dept",
})
```

```python
df["Monthly_Salary"] = df["Monthly_Salary"].fillna(df["Monthly_Salary"].mean())
df["Dept"] = df["Dept"].fillna("Unknown")
```

```python
df["Monthly_Salary"] = df["Monthly_Salary"]/12
df["Level"] = df["Monthly_Salary"].apply(lambda x: "Senior" if x >=80000 else ("Mid" if x >=50000 else "Junior"))
```

```python
df = df.sort_values("Monthly_Salary",ascending = False).reset_index(drop = True)
print(f"Cleaned Data: {df.shape[0]} rows, {df.shape[1]} columns")
print(df.head())
```

```python
# Load
df.to_csv("cleaned_employees.csv",index = False)
print("ETL Complete! File saved.") 
```
